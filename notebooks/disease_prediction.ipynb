```python
import tensorflow as tf

data_dir = "dataset/plantvillage dataset/color"

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(224, 224),
    batch_size=32)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(224, 224),
    batch_size=32)
```

```python
num_classes = len(train_ds.class_names)
print(num_classes)  # should print 38

model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(32, 3, activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(num_classes)
])

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy'])
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)
```

```python
model.save('plant_disease_model_v1.keras')
```

```python
import os
import tensorflow as tf

# Ensure the dataset is present by re-downloading and re-unzipping
!kaggle datasets download -d abdallahalidev/plantvillage-dataset -q
!unzip -q -o plantvillage-dataset.zip -d dataset

data_dir = "dataset/plantvillage dataset/color"

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(224, 224),
    batch_size=32)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(224, 224),
    batch_size=32)

num_classes = len(train_ds.class_names)
# This rebuilds train_ds, val_ds, and num_classes — quick, since it's just reading files that are already on disk (assuming step 2 confirmed they're there).
```

```python
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
```

```python
base_model = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights='imagenet',
    input_shape=(224, 224, 3)
)
base_model.trainable = False  # freeze it for phase 1

inputs = tf.keras.Input(shape=(224, 224, 3))
x = tf.keras.applications.efficientnet.preprocess_input(inputs)
x = base_model(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.3)(x)
outputs = tf.keras.layers.Dense(num_classes)(x)
model2 = tf.keras.Model(inputs, outputs)

model2.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy'])
```

```python
history2 = model2.fit(train_ds, validation_data=val_ds, epochs=5)
```

```python
model2.save('plant_disease_phase1.keras')
```

```python
base_model.trainable = True

# Freeze all layers except the last ~30
for layer in base_model.layers[:-30]:
    layer.trainable = False

model2.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),  # much lower LR
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy'])
```

```python
history_fine = model2.fit(train_ds, validation_data=val_ds, epochs=5)
```
